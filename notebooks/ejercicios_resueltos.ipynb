```python
# Ejercicio 1:  Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias
# de cada letra en la cadena. Los espacios no deben ser considerados.
 
def contar_frecuencia_letras(texto):
    """
    Recibe una cadena de texto y devuelve un diccionario
    con la frecuencia de cada letra.
    Los espacios no se tienen en cuenta.
    """
    
    frecuencias = {}
    
    for letra in texto:
        if letra != " ":
            if letra in frecuencias:
                frecuencias[letra] += 1
            else:
                frecuencias[letra] = 1
    
    return frecuencias
```

```python
texto = "Hola mundo"

resultado = contar_frecuencia_letras(texto)

print(resultado)
```

```python
# Ejercicio 2:  Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map()

def duplicar_valores(lista_numeros):
    """
    Recibe una lista de números y devuelve
    una nueva lista con el doble de cada valor.
    """
    
    resultado = list(map(lambda x: x * 2, lista_numeros))
    
    return resultado
```

```python
numeros = [1, 2, 3, 4]

resultado = duplicar_valores(numeros)

print(resultado)
```

```python
# Ejercicio 3:  Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe
# devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.
 
def buscar_palabras(lista_palabras, palabra_objetivo):
    """
    Devuelve una lista con las palabras que contienen
    la palabra objetivo.
    """
    
    resultado = []
    
    for palabra in lista_palabras:
        if palabra_objetivo in palabra:
            resultado.append(palabra)
    
    return resultado
```

```python
palabras = ["casa", "casita", "perro", "casona"]

resultado = buscar_palabras(palabras, "casa")

print(resultado)
```

```python
# Ejercicio 4:  Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map()
 
def calcular_diferencia(lista1, lista2):
    """
    Calcula la diferencia entre los valores
    de dos listas usando map().
    """
    
    resultado = list(map(lambda x, y: x - y, lista1, lista2))
    
    return resultado
```

```python
lista_a = [10, 20, 30]
lista_b = [1, 2, 3]

resultado = calcular_diferencia(lista_a, lista_b)

print(resultado)
```

```python
# Ejercicio 5: Ecribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado, que por
# defecto es 5. La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual
# que nota aprobado. Si es así, el estado será "aprobado", de lo contrario, será "suspenso". La función debe devolver
# una tupla que contenga la media y el estado.
 
def calcular_media_estado(lista_numeros, nota_aprobado=5):
    """
    Calcula la media de una lista de números y devuelve
    una tupla con la media y el estado.
    """
    
    media = sum(lista_numeros) / len(lista_numeros)
    
    if media >= nota_aprobado:
        estado = "aprobado"
    else:
        estado = "suspenso"
    
    return (media, estado)

```

```python
notas = [7, 8, 6, 9]

resultado = calcular_media_estado(notas)

print(resultado)
```

```python
# Ejercicio 6: Escribe una función que calcule el factorial de un número de manera recursiva.
 
def calcular_factorial(numero):
    """
    Calcula el factorial de un número
    usando recursividad.
    """
    
    if numero == 0 or numero == 1:
        return 1
    
    return numero * calcular_factorial(numero - 1)
```

```python
resultado = calcular_factorial(5)

print(resultado)
```

```python
# Ejercicio 7: Genera una función que convierta una lista de tuplas a una lista de strings. Usa la función map()
 
def convertir_tuplas_a_strings(lista_tuplas):
    """
    Convierte una lista de tuplas
    en una lista de strings usando map().
    """
    
    resultado = list(map(lambda tupla: " ".join(map(str, tupla)), lista_tuplas))
    
    return resultado
```

```python
tuplas = [("Hola", "mundo"), ("Python", 3)]

resultado = convertir_tuplas_a_strings(tuplas)

print(resultado)
```

```python
# Ejercicio 8: Escribe un programa que pida al usuario dos números e intente dividirlos. Si el usuario ingresa un valor no numérico
# o intenta dividir por cero, maneja esas excepciones de manera adecuada. Asegúrate de mostrar un mensaje
# indicando si la división fue exitosa o no.
 
def dividir_numeros():
    """
    Pide dos números al usuario e intenta dividirlos,
    manejando posibles excepciones.
    """
    
    try:
        numero1 = float(input("Introduce el primer número: "))
        numero2 = float(input("Introduce el segundo número: "))
        
        resultado = numero1 / numero2
        
        print(f"La división es: {resultado}")
        print("La división fue exitosa.")
    
    except ValueError:
        print("Error: Debes introducir valores numéricos.")
    
    except ZeroDivisionError:
        print("Error: No se puede dividir entre cero.")
    
    finally:
        print("Programa finalizado.")
```

```python
dividir_numeros()
```

```python
# Ejercicio 9: Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista
# excluyendo ciertas mascotas prohibidas en España. La lista de mascotas a excluir es ["Mapache", "Tigre",
# "Serpiente Pitón", "Cocodrilo", "Oso"].Usa la función filter()
 
def filtrar_mascotas(lista_mascotas):
    """
    Devuelve una nueva lista excluyendo
    las mascotas prohibidas usando filter().
    """
    
    mascotas_prohibidas = [
        "Mapache",
        "Tigre",
        "Serpiente Pitón",
        "Cocodrilo",
        "Oso"
    ]
    
    resultado = list(
        filter(
            lambda mascota: mascota not in mascotas_prohibidas,
            lista_mascotas
        )
    )
    
    return resultado
 
```

```python
mascotas = ["Perro", "Gato", "Tigre", "Oso", "Canario"]

resultado = filtrar_mascotas(mascotas)

print(resultado)
```

```python
# Ejercicio 10: Escribe una función que reciba una lista de números y calcule su promedio. Si la lista está vacía, lanza una
# excepción personalizada y maneja el error adecuadamente.
 
class ListaVaciaError(Exception):
    """
    Excepción personalizada para listas vacías.
    """
    pass


def calcular_promedio(lista_numeros):
    """
    Calcula el promedio de una lista de números.
    Lanza una excepción si la lista está vacía.
    """
    
    try:
        if len(lista_numeros) == 0:
            raise ListaVaciaError("La lista está vacía.")
        
        promedio = sum(lista_numeros) / len(lista_numeros)
        
        return promedio
    
    except ListaVaciaError as error:
        print(f"Error: {error}")
```

```python
numeros = [5, 7, 9]

resultado = calcular_promedio(numeros)

print(resultado)
```

```python
# Ejercicio 11: Escribe un programa que pida al usuario que introduzca su edad. Si el usuario ingresa un valor no numérico o un
# valor fuera del rango esperado (por ejemplo, menor que 0 o mayor que 120, maneja las excepciones
# adecuadamente.
 
def validar_edad():
    """
    Pide al usuario su edad y maneja
    posibles errores de entrada.
    """
    
    try:
        edad = int(input("Introduce tu edad: "))
        
        if edad < 0 or edad > 120:
            raise ValueError("La edad debe estar entre 0 y 120.")
        
        print(f"Edad registrada correctamente: {edad} años.")
    
    except ValueError as error:
        print(f"Error: {error}")
    
    finally:
        print("Programa finalizado.")
```

```python
# Ejercicio 12: Genera una función que al recibir una frase devuelva una lista con la longitud de cada palabra. Usa la función map()
 
def longitud_palabras(frase):
    """
    Devuelve una lista con la longitud
    de cada palabra usando map().
    """
    
    palabras = frase.split()
    
    resultado = list(map(len, palabras))
    
    return resultado
```

```python
frase = "Estoy aprendiendo Python"

resultado = longitud_palabras(frase)

print(resultado)
```

```python
# Ejercicio 13: Genera una función la cual, para un conjunto de caracteres, devuelva una lista de tuplas con cada letra en
# mayúsculas y minúsculas. Las letras no pueden estar repetidas .Usa la función map()
 
def convertir_mayusculas_minusculas(caracteres):
    """
    Recibe una lista con una palabra y devuelve una lista
    de tuplas con cada letra en mayúscula y minúscula.
    Las letras no se repiten.
    """

    palabra = caracteres[0]

    letras_unicas = list(dict.fromkeys(palabra))

    resultado = list(
        map(
            lambda letra: (letra.upper(), letra.lower()),
            letras_unicas
        )
    )

    return resultado
```

```python
caracteres = ["python"]

resultado = convertir_mayusculas_minusculas(caracteres)

print(resultado)
```

```python
# Ejercicio 14: Crea una función que retorne las palabras de una lista de palabras que comience con una letra en especifico. Usa la
# función filter()
 
def filtrar_palabras_por_letra(lista_palabras, letra):
    """
    Devuelve las palabras que comienzan
    por una letra específica usando filter().
    """

    resultado = list(
        filter(
            lambda palabra: palabra.startswith(letra),
            lista_palabras
        )
    )

    return resultado
```

```python
palabras = ["mesa", "manzana", "perro", "mono"]

resultado = filtrar_palabras_por_letra(palabras, "m")

print(resultado)
```

```python
# Ejercicio 16: Escribe una función que tome una cadena de texto y un número entero n como parámetros y devuelva una lista de
# todas las palabras que sean más largas que n. Usa la función filter()

def palabras_mas_largas(texto, n):
    """
    Devuelve una lista con las palabras
    cuya longitud es mayor que n usando filter().
    """

    palabras = texto.split()

    resultado = list(
        filter(
            lambda palabra: len(palabra) > n,
            palabras
        )
    )

    return resultado
```

```python
frase = "Python es un lenguaje muy interesante"

resultado = palabras_mas_largas(frase, 4)

print(resultado)
```

```python
# Ejercicio 17:  Crea una función que tome una lista de dígitos y devuelva el número correspondiente. Por ejemplo, [5,7,2]
# corresponde al número quinientos setenta y dos 572. Usa la función reduce()
 
from functools import reduce


def convertir_lista_a_numero(lista_digitos):
    """
    Convierte una lista de dígitos
    en un número usando reduce().
    """

    numero = reduce(
        lambda acumulador, digito: acumulador * 10 + digito,
        lista_digitos
    )

    return numero
```

```python
digitos = [5, 7, 2]

resultado = convertir_lista_a_numero(digitos)

print(resultado)
```

```python
# Ejercicio 18: Escribe un programa en Python que cree una lista de diccionarios que contenga información de estudiantes
# (nombre, edad, calificación) y use la función filter para extraer a los estudiantes con una calificación mayor o igual a
# 90. Usa la función filter()
 
def filtrar_estudiantes_destacados(lista_estudiantes):
    """
    Devuelve los estudiantes con una
    calificación mayor o igual a 90 usando filter().
    """

    resultado = list(
        filter(
            lambda estudiante: estudiante["calificacion"] >= 90,
            lista_estudiantes
        )
    )

    return resultado
```

```python
estudiantes = [
        {"nombre": "Ana", "edad": 20, "calificacion": 95},
        {"nombre": "Luis", "edad": 21, "calificacion": 82},
        {"nombre": "Marta", "edad": 22, "calificacion": 90}
    ]

resultado = filtrar_estudiantes_destacados(estudiantes)

print(resultado)
```

```python
# Ejercicio 19: Crea una función lambda que filtre los números impares de una lista dada.
 
filtrar_impares = lambda lista: list(
    filter(
        lambda numero: numero % 2 != 0,
        lista
    )
)
```

```python
numeros = [1, 2, 3, 4, 5, 6]

resultado = filtrar_impares(numeros)

print(resultado)
```

```python
# Ejercicio 20:  Para una lista con elementos tipo integer y string obtén una nueva lista sólo con los valores int. Usa la función
# filter()
 
def filtrar_enteros(lista):
    """
    Devuelve una nueva lista solo con
    los valores de tipo int usando filter().
    """

    resultado = list(
        filter(
            lambda elemento: isinstance(elemento, int),
            lista
        )
    )

    return resultado
```

```python
datos = [1, "hola", 3, "python", 5]

resultado = filtrar_enteros(datos)

print(resultado)
```
